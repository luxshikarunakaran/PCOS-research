```python
# =============================================================================
#  PCOS CLASSIFICATION — SIAMESE / TRIPLET / CAPSULE NETWORKS
#  Stratified 5-Fold CV  +  full preprocessing  +  full metric suite
#  +  full visualisation suite  +  unseen-data prediction & verification
#
#  Files expected:
#     Training_data.csv  -> Cohort A : clinical  (541 x 45)  target "PCOS (Y/N)"
#     Testing_data.csv   -> Cohort B : lifestyle (173 x 36)  target "PCOS"
#
#  IMPORTANT — READ THIS FIRST
#  ---------------------------------------------------------------------------
#  The two files are NOT a train/test split of one dataset. They share ZERO
#  columns (clinical hormone panel vs. lifestyle survey). A model fitted on one
#  cannot be scored on the other. So this script runs the COMPLETE pipeline
#  independently on each cohort, and for each cohort it carves its own
#  stratified HOLD-OUT set that is never touched until the final verification
#  stage. That hold-out is the "new data" used to prove the models work.
#
#  Layout: 16 cells, marked "# %%". Paste cell-by-cell into Colab, or run the
#  whole file with `python pcos_deep_metric_capsnet.py`.
# =============================================================================

# %% ==========================  CELL 0 — SETUP  ==============================
# !pip install torch scikit-learn pandas numpy matplotlib seaborn --quiet

import os, sys, json, math, time, random, warnings, itertools
from collections import Counter, defaultdict

import numpy as np
import pandas as pd
import matplotlib
matplotlib.use("Agg")                      # comment out in Colab for inline plots
import matplotlib.pyplot as plt
import seaborn as sns

import torch
import torch.nn as nn
import torch.nn.functional as F
from torch.utils.data import Dataset, DataLoader

from sklearn.model_selection import StratifiedKFold, train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
from sklearn.manifold import TSNE
from sklearn.neighbors import KNeighborsClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.dummy import DummyClassifier
from sklearn.metrics import (
    accuracy_score, balanced_accuracy_score, precision_score, recall_score,
    f1_score, fbeta_score, roc_auc_score, average_precision_score,
    matthews_corrcoef, cohen_kappa_score, log_loss, brier_score_loss,
    confusion_matrix, roc_curve, precision_recall_curve, silhouette_score,
    classification_report,
)
from sklearn.calibration import calibration_curve

warnings.filterwarnings("ignore")

# ----------------------------- configuration --------------------------------
CFG = dict(
    SEED            = 42,
    N_FOLDS         = 5,
    HOLDOUT_FRAC    = 0.20,     # never seen during CV or final training
    EMB_DIM         = 32,       # Siamese / Triplet embedding size
    HIDDEN          = (128, 64),
    DROPOUT         = 0.30,
    LR              = 1e-3,
    WEIGHT_DECAY    = 1e-4,
    EPOCHS_METRIC   = 120,      # Siamese / Triplet
    EPOCHS_CAPS     = 100,      # CapsNet
    BATCH           = 64,
    MARGIN_CONTRAST = 1.0,
    MARGIN_TRIPLET  = 0.5,
    SOFTMAX_TEMP    = 0.20,     # temperature for centroid-distance -> probability
    ROUTING_ITERS   = 3,
    CAPS_PRIM_DIM   = 8,
    CAPS_OUT_DIM    = 16,
    RECON_WEIGHT    = 5e-4,
    THRESHOLD       = 0.50,
)

FIGDIR = "figs"
OUTDIR = "outputs"
os.makedirs(FIGDIR, exist_ok=True)
os.makedirs(OUTDIR, exist_ok=True)

DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")


def set_seed(seed=CFG["SEED"]):
    random.seed(seed); np.random.seed(seed)
    torch.manual_seed(seed); torch.cuda.manual_seed_all(seed)
    os.environ["PYTHONHASHSEED"] = str(seed)


set_seed()
sns.set_theme(style="whitegrid", context="notebook")
plt.rcParams["figure.dpi"] = 110
plt.rcParams["savefig.bbox"] = "tight"

_FIG_N = [0]


def savefig(name):
    """Save + close the current figure. In Colab replace close() with show()."""
    _FIG_N[0] += 1
    path = os.path.join(FIGDIR, f"{_FIG_N[0]:03d}_{name}.png")
    plt.savefig(path)
    plt.close()
    return path


def banner(txt, ch="="):
    print("\n" + ch * 78 + f"\n{txt}\n" + ch * 78)


_trapz = getattr(np, "trapezoid", None) or np.trapz   # numpy 1.x / 2.x compat

print(f"torch {torch.__version__} | device={DEVICE} | seed={CFG['SEED']}")


# %% ======================  CELL 1 — LOAD + RAW AUDIT  =======================
# Adjust these two paths for your environment.
PATH_A = "/content/drive/MyDrive/PCOD/Training data.csv"     # clinical cohort
PATH_B = "/content/drive/MyDrive/PCOD/Testing data.csv"      # lifestyle-survey cohort

for _p, _alt in [("PATH_A", "/content/drive/MyDrive/PCOD/Training data.csv"),
                 ("PATH_B", "/content/drive/MyDrive/PCOD/Testing data.csv")]:
    if not os.path.exists(globals()[_p]) and os.path.exists(_alt):
        globals()[_p] = _alt

raw_A = pd.read_csv(PATH_A)
raw_B = pd.read_csv(PATH_B)

banner("CELL 1 — RAW DATA AUDIT")
print(f"Cohort A (clinical)  : {raw_A.shape[0]} rows x {raw_A.shape[1]} cols")
print(f"Cohort B (lifestyle) : {raw_B.shape[0]} rows x {raw_B.shape[1]} cols")
print(f"Columns shared between the two files: {len(set(raw_A.columns) & set(raw_B.columns))}")
print("  -> they are two different datasets, not a train/test split.\n")


def audit(df, name):
    print(f"--- {name} ---")
    rep = pd.DataFrame({
        "dtype":    df.dtypes.astype(str),
        "n_unique": df.nunique(),
        "n_missing": df.isna().sum(),
        "pct_missing": (df.isna().mean() * 100).round(2),
    })
    print(rep.to_string())
    print(f"duplicate rows: {df.duplicated().sum()}\n")
    return rep


audit_A = audit(raw_A, "Cohort A raw")
audit_B = audit(raw_B, "Cohort B raw")


def coerce_numeric_report(df, cols):
    """Show which values block a column from parsing as numeric."""
    for c in cols:
        if c not in df.columns:
            continue
        s = df[c].astype(str)
        bad = s[pd.to_numeric(s, errors="coerce").isna() & df[c].notna()]
        if len(bad):
            print(f"  !! '{c}' has {len(bad)} non-numeric token(s): "
                  f"{sorted(bad.unique())[:6]}")


print("Columns that LOOK numeric but silently parse as text:")
coerce_numeric_report(raw_A, [c for c in raw_A.columns
                              if raw_A[c].dtype.kind not in "if"])


# %% ====================  CELL 2 — CLEAN COHORT A (clinical)  ================
banner("CELL 2 — CLEANING COHORT A (clinical)")

A = raw_A.copy()
A.columns = [c.strip() for c in A.columns]

TARGET_A = "PCOS (Y/N)"

# 1. drop identifier columns + the stray unnamed column (539/541 empty)
drop_A = [c for c in ["Sl. No", "Patient File No.", "Unnamed: 44"] if c in A.columns]
A = A.drop(columns=drop_A)
print(f"dropped identifier/junk columns: {drop_A}")

# 2. force numeric on the two poisoned columns (AMH has literal 'a',
#    II beta-HCG has '1.99.') — this is the single most common silent bug here
for c in A.columns:
    if c != TARGET_A and A[c].dtype.kind not in "if":
        before = A[c].isna().sum()
        A[c] = pd.to_numeric(A[c], errors="coerce")
        print(f"  coerced '{c}' -> numeric, new NaNs introduced: "
              f"{A[c].isna().sum() - before}")

# 3. target
A[TARGET_A] = pd.to_numeric(A[TARGET_A], errors="coerce").astype(int)

# 4. drop exact duplicate rows
n0 = len(A); A = A.drop_duplicates().reset_index(drop=True)
print(f"duplicates removed: {n0 - len(A)}")

# 5. physiologically impossible values -> NaN, then median-impute
sanity_A = {
    "Age (yrs)": (10, 70), "Weight (Kg)": (25, 200), "Height(Cm)": (120, 210),
    "BMI": (10, 70), "Pulse rate(bpm)": (35, 200), "RR (breaths/min)": (8, 40),
    "Hb(g/dl)": (5, 20), "Cycle length(days)": (1, 60),
    "BP _Systolic (mmHg)": (70, 220), "BP _Diastolic (mmHg)": (40, 140),
}
for c, (lo, hi) in sanity_A.items():
    if c in A.columns:
        bad = (~A[c].between(lo, hi)) & A[c].notna()
        if bad.sum():
            print(f"  '{c}': {bad.sum()} out-of-range value(s) -> NaN")
            A.loc[bad, c] = np.nan

# 6. binary Y/N columns -> keep as 0/1, everything else numeric
bin_A = [c for c in A.columns
         if c != TARGET_A and set(A[c].dropna().unique()) <= {0, 1}]
num_A = [c for c in A.columns if c not in bin_A + [TARGET_A]]

# 7. median impute (fold-safe: medians are recomputed inside CV via the
#    Preprocessor class in CELL 5; this pass only makes the frame plottable)
A[num_A] = A[num_A].fillna(A[num_A].median())
A[bin_A] = A[bin_A].fillna(0)

# 8. engineered clinical ratios (Rotterdam-flavoured)
if {"Follicle No. (L)", "Follicle No. (R)"} <= set(A.columns):
    A["Follicle_total"] = A["Follicle No. (L)"] + A["Follicle No. (R)"]
    A["Follicle_asym"] = (A["Follicle No. (L)"] - A["Follicle No. (R)"]).abs()
if {"Avg. F size (L) (mm)", "Avg. F size (R) (mm)"} <= set(A.columns):
    A["Fsize_mean"] = (A["Avg. F size (L) (mm)"] + A["Avg. F size (R) (mm)"]) / 2
if {"Weight (Kg)", "Height(Cm)"} <= set(A.columns):
    A["BMI_recomputed"] = A["Weight (Kg)"] / (A["Height(Cm)"] / 100) ** 2

X_A = A.drop(columns=[TARGET_A])
y_A = A[TARGET_A].values.astype(int)
print(f"\nCohort A ready : X={X_A.shape}  y balance={Counter(y_A)}")


# %% ===================  CELL 3 — CLEAN COHORT B (lifestyle)  ================
banner("CELL 3 — CLEANING COHORT B (lifestyle survey)")

B = raw_B.copy()
B.columns = [c.strip() for c in B.columns]
TARGET_B = "PCOS"

# 1. TARGET LEAKAGE: 'PCOS_Medication' is only ever filled for diagnosed
#    patients — it encodes the label. Dropping it is mandatory.
if "PCOS_Medication" in B.columns:
    B = B.drop(columns=["PCOS_Medication"])
    print("dropped 'PCOS_Medication' (direct target leakage)")

# 2. one row has an unusable target: 'No, Yes, not diagnosed by a doctor'
mask_bad = ~B[TARGET_B].astype(str).str.strip().str.lower().isin(["yes", "no"])
print(f"ambiguous target rows dropped: {int(mask_bad.sum())} "
      f"-> {B.loc[mask_bad, TARGET_B].tolist()}")
B = B[~mask_bad].reset_index(drop=True)
B[TARGET_B] = (B[TARGET_B].astype(str).str.strip().str.lower() == "yes").astype(int)

# 3. 'Height_ft' actually stores CENTIMETRES (values ~150-180). Rename so the
#    feature name stops lying, and derive BMI from it.
if "Height_ft" in B.columns:
    B = B.rename(columns={"Height_ft": "Height_cm"})
    print(f"renamed Height_ft -> Height_cm (observed range "
          f"{B['Height_cm'].min():.0f}-{B['Height_cm'].max():.0f}, clearly cm)")
    if "Weight_kg" in B.columns:
        B["BMI_derived"] = B["Weight_kg"] / (B["Height_cm"] / 100) ** 2

# 4. The symptom columns are MULTI-SELECT survey answers, so a plain Yes/No
#    map loses information. Encode them on a 3-level clinical-certainty scale:
#       0.0 = No      0.5 = self-reported / mixed answer      1.0 = doctor-confirmed
def symptom_scale(s):
    v = s.astype(str).str.strip().str.lower()
    out = pd.Series(0.5, index=s.index, dtype=float)
    out[v.eq("no")] = 0.0
    out[v.eq("yes") | v.str.contains("yes, diagnosed by a doctor", na=False)] = 1.0
    out[~v.str.contains("yes", na=False) & ~v.eq("no")] = 0.0
    return out


symptom_cols = ["Family_History_PCOS", "Menstrual_Irregularity", "Hormonal_Imbalance",
                "Hyperandrogenism", "Hirsutism", "Mental_Health", "Conception_Difficulty",
                "Insulin_Resistance", "Diabetes", "Childhood_Trauma",
                "Cardiovascular_Disease", "Vegetarian", "Smoking"]
for c in symptom_cols:
    if c in B.columns:
        B[c] = symptom_scale(B[c])
        lv = sorted(B[c].unique())
        print(f"  '{c}' -> certainty scale {lv}")

# 5. ORDERED categoricals -> ordinal codes. Ordinal (not one-hot) keeps the
#    ranking AND keeps dimensionality low, which matters at n=172.
ORDINALS = {
    "Age": {"less than 20": 0, "20-25": 1, "25-30": 2, "30-35": 3,
            "35-44": 4, "45 and above": 5},
    "Exercise_Frequency": {"never": 0, "rarely": 1, "1-2 times a week": 2,
                           "3-4 times a week": 3, "daily": 4},
    "Exercise_Duration": {"not applicable": 0, "less than 30 minutes": 1,
                          "30 minutes": 2, "more than 30 minutes": 3,
                          "30 minutes to 1 hour": 3},
    "Sleep_Hours": {"less than 6 hours": 0, "6-8 hours": 1,
                    "9-12 hours": 2, "more than 12 hours": 3},
    "Exercise_Benefit": {"not at all": 0, "not much": 1, "somewhat": 2,
                         "yes significantly": 3},
    "Stress_Level": {"no": 0, "yes": 1},
}
for col, mapping in ORDINALS.items():
    if col not in B.columns:
        continue
    v = B[col].astype(str).str.strip().str.lower()
    coded = v.map(mapping)
    if coded.isna().any():
        unseen = sorted(v[coded.isna()].unique())
        raise ValueError(f"'{col}' has unmapped level(s) {unseen} — extend ORDINALS")
    B[col] = coded.astype(float)
    print(f"  '{col}' -> ordinal 0..{int(B[col].max())}")

# 6. 'Exercise_Type' is multi-select free text with 12 combinations. One-hot
#    would spend 11 columns on 172 rows; multi-hot on the 5 base modalities
#    spends 5 and generalises far better.
if "Exercise_Type" in B.columns:
    et = B["Exercise_Type"].astype(str).str.lower()
    modalities = {
        "Ex_Cardio":      "cardio",
        "Ex_Strength":    "strength training",
        "Ex_Flexibility": "flexibility and balance",
        "Ex_HIIT":        "high-intensity interval",
        "Ex_None":        "no exercise",
    }
    for newc, key in modalities.items():
        B[newc] = et.str.contains(key, regex=False).astype(float)
    B["Ex_ModalityCount"] = B[[c for c in modalities if c != "Ex_None"]].sum(axis=1)
    B = B.drop(columns=["Exercise_Type"])
    print(f"  'Exercise_Type' -> multi-hot {list(modalities)} + Ex_ModalityCount")

# 6b. remaining nominal text (Marital_Status) -> one-hot
nominal = [c for c in B.columns if B[c].dtype.kind not in "if" and c != TARGET_B]
print(f"one-hot encoded nominal columns: {nominal}")
if nominal:
    B = pd.get_dummies(B, columns=nominal, drop_first=True, dtype=float)

# 7. diet frequency columns are already 0-7 integer scales — leave as ordinal
B = B.drop_duplicates().reset_index(drop=True)
B = B.fillna(B.median(numeric_only=True))

X_B = B.drop(columns=[TARGET_B])
y_B = B[TARGET_B].values.astype(int)
print(f"\nCohort B ready : X={X_B.shape}  y balance={Counter(y_B)}")
print(f"  positive rate = {y_B.mean():.3f}  (imbalanced — balanced accuracy, "
      f"MCC and PR-AUC will matter more than raw accuracy)")

COHORTS = {
    "A_clinical":  dict(X=X_A, y=y_A, label="Cohort A — clinical (Training_data.csv)"),
    "B_lifestyle": dict(X=X_B, y=y_B, label="Cohort B — lifestyle (Testing_data.csv)"),
}


# %% =================  CELL 4 — EXPLORATORY DATA VISUALISATION  ==============
banner("CELL 4 — EDA VISUALISATIONS")


def eda_visuals(X, y, tag, label, raw_df=None, top_k=12):
    figs = []

    # 1. class balance
    plt.figure(figsize=(4.5, 3.6))
    ax = sns.countplot(x=pd.Series(y).map({0: "No PCOS", 1: "PCOS"}),
                       hue=pd.Series(y).map({0: "No PCOS", 1: "PCOS"}),
                       palette="Set2", legend=False)
    for p in ax.patches:
        ax.annotate(f"{int(p.get_height())}\n({p.get_height()/len(y):.1%})",
                    (p.get_x() + p.get_width()/2, p.get_height()),
                    ha="center", va="bottom", fontsize=9)
    plt.title(f"{label}\nclass balance"); plt.ylabel("count"); plt.xlabel("")
    figs.append(savefig(f"{tag}_class_balance"))

    # 2. missingness (on the raw frame, before imputation)
    if raw_df is not None:
        miss = raw_df.isna().mean().sort_values(ascending=False)
        miss = miss[miss > 0]
        if len(miss):
            plt.figure(figsize=(6, max(2.2, 0.3 * len(miss))))
            sns.barplot(x=miss.values * 100, y=miss.index, hue=miss.index,
                        palette="rocket", legend=False)
            plt.xlabel("% missing"); plt.title(f"{label}\nmissingness (raw)")
            figs.append(savefig(f"{tag}_missingness"))

    # 3. correlation heatmap
    corr = X.corr(numeric_only=True)
    plt.figure(figsize=(min(16, 0.28 * len(corr) + 4),
                        min(14, 0.28 * len(corr) + 3)))
    sns.heatmap(corr, cmap="coolwarm", center=0, square=True,
                cbar_kws=dict(shrink=0.6), xticklabels=True, yticklabels=True)
    plt.xticks(fontsize=6, rotation=90); plt.yticks(fontsize=6)
    plt.title(f"{label}\nfeature correlation")
    figs.append(savefig(f"{tag}_corr_heatmap"))

    # 4. |point-biserial correlation with target| — ranked
    tgt_corr = X.apply(lambda c: np.corrcoef(c, y)[0, 1]
                       if c.std() > 0 else 0.0).sort_values(key=np.abs,
                                                            ascending=False)
    plt.figure(figsize=(6, max(3, 0.26 * len(tgt_corr))))
    colors = ["#d1495b" if v > 0 else "#00798c" for v in tgt_corr.values]
    plt.barh(range(len(tgt_corr)), tgt_corr.values, color=colors)
    plt.yticks(range(len(tgt_corr)), tgt_corr.index, fontsize=6)
    plt.gca().invert_yaxis(); plt.axvline(0, color="k", lw=0.8)
    plt.xlabel("correlation with PCOS label")
    plt.title(f"{label}\nfeature–target correlation (ranked)")
    figs.append(savefig(f"{tag}_target_corr"))

    # 5. distribution of the top-k discriminative features, split by class
    top = tgt_corr.head(top_k).index.tolist()
    ncol, nrow = 4, int(np.ceil(len(top) / 4))
    plt.figure(figsize=(4 * ncol, 2.9 * nrow))
    for i, c in enumerate(top, 1):
        plt.subplot(nrow, ncol, i)
        if X[c].nunique() <= 6:
            sns.countplot(x=X[c], hue=pd.Series(y).map({0: "No", 1: "Yes"}),
                          palette="Set2")
        else:
            for cls, col in [(0, "#00798c"), (1, "#d1495b")]:
                sns.kdeplot(X.loc[y == cls, c], fill=True, alpha=.35,
                            color=col, label="Yes" if cls else "No",
                            warn_singular=False)
        plt.title(c, fontsize=8); plt.xlabel(""); plt.legend(fontsize=6)
    plt.suptitle(f"{label}\ntop-{top_k} features by |corr| — distribution by class",
                 y=1.01)
    plt.tight_layout()
    figs.append(savefig(f"{tag}_top_feature_dists"))

    # 6. boxplots (standardised) for the same top-k — outlier view
    Z = pd.DataFrame(StandardScaler().fit_transform(X[top]), columns=top)
    Z["PCOS"] = y
    Zm = Z.melt(id_vars="PCOS", var_name="feature", value_name="z")
    plt.figure(figsize=(1.1 * len(top) + 3, 4.2))
    sns.boxplot(data=Zm, x="feature", y="z",
                hue=Zm["PCOS"].map({0: "No", 1: "Yes"}), palette="Set2")
    plt.xticks(rotation=40, ha="right", fontsize=7)
    plt.title(f"{label}\nstandardised distributions + outliers")
    figs.append(savefig(f"{tag}_boxplots"))

    # 7. PCA (2D) of the raw standardised space
    Xs = StandardScaler().fit_transform(X)
    pca = PCA(n_components=min(10, Xs.shape[1]), random_state=CFG["SEED"])
    P = pca.fit_transform(Xs)
    plt.figure(figsize=(10, 4))
    plt.subplot(1, 2, 1)
    for cls, col, nm in [(0, "#00798c", "No PCOS"), (1, "#d1495b", "PCOS")]:
        plt.scatter(P[y == cls, 0], P[y == cls, 1], s=16, alpha=.7,
                    c=col, label=nm)
    plt.xlabel(f"PC1 ({pca.explained_variance_ratio_[0]:.1%})")
    plt.ylabel(f"PC2 ({pca.explained_variance_ratio_[1]:.1%})")
    plt.legend(); plt.title("PCA — input space")
    plt.subplot(1, 2, 2)
    cum = np.cumsum(pca.explained_variance_ratio_)
    plt.plot(range(1, len(cum) + 1), cum, "o-")
    plt.axhline(.9, ls="--", c="r"); plt.xlabel("components")
    plt.ylabel("cumulative explained variance"); plt.title("Scree")
    plt.suptitle(label)
    plt.tight_layout()
    figs.append(savefig(f"{tag}_pca_input"))

    # 8. t-SNE of the raw standardised space (baseline for comparison against
    #    the learned Siamese/Triplet embeddings later)
    per = max(5, min(30, (len(Xs) - 1) // 3))
    T = TSNE(n_components=2, perplexity=per, init="pca",
             random_state=CFG["SEED"]).fit_transform(Xs)
    plt.figure(figsize=(5, 4.4))
    for cls, col, nm in [(0, "#00798c", "No PCOS"), (1, "#d1495b", "PCOS")]:
        plt.scatter(T[y == cls, 0], T[y == cls, 1], s=18, alpha=.75,
                    c=col, label=nm)
    sil = silhouette_score(Xs, y)
    plt.legend(); plt.title(f"{label}\nt-SNE of INPUT space "
                            f"(class silhouette = {sil:.3f})")
    figs.append(savefig(f"{tag}_tsne_input"))

    print(f"[{tag}] {len(figs)} EDA figures written; input-space class "
          f"silhouette = {sil:.3f}")
    return dict(target_corr=tgt_corr, input_silhouette=sil, figs=figs)


eda_results = {}
eda_results["A_clinical"] = eda_visuals(X_A, y_A, "A", COHORTS["A_clinical"]["label"],
                                        raw_df=raw_A)
eda_results["B_lifestyle"] = eda_visuals(X_B, y_B, "B", COHORTS["B_lifestyle"]["label"],
                                         raw_df=raw_B)


# %% ==========  CELL 5 — LEAK-FREE PREPROCESSOR + HOLD-OUT SPLIT  ===========
banner("CELL 5 — PREPROCESSING PIPELINE + HOLD-OUT SPLIT")


class Preprocessor:
    """Fit ONLY on training rows. Imputation medians, IQR clip bounds and the
    scaler are all learned from the training fold, then applied unchanged to
    validation / hold-out / brand-new rows. This is what stops CV scores from
    being optimistically biased."""

    def __init__(self, clip_iqr=3.0, var_thresh=1e-8):
        self.clip_iqr = clip_iqr
        self.var_thresh = var_thresh

    def fit(self, X: pd.DataFrame):
        self.columns_ = list(X.columns)
        self.median_ = X.median(numeric_only=True)
        q1, q3 = X.quantile(.25), X.quantile(.75)
        iqr = (q3 - q1).replace(0, np.nan)
        self.lo_ = (q1 - self.clip_iqr * iqr).fillna(X.min())
        self.hi_ = (q3 + self.clip_iqr * iqr).fillna(X.max())
        Xc = X.fillna(self.median_).clip(self.lo_, self.hi_, axis=1)
        self.keep_ = [c for c in self.columns_ if Xc[c].var() > self.var_thresh]
        self.scaler_ = StandardScaler().fit(Xc[self.keep_].values)
        return self

    def transform(self, X: pd.DataFrame) -> np.ndarray:
        X = X.reindex(columns=self.columns_)          # tolerate column reorder
        Xc = X.fillna(self.median_).clip(self.lo_, self.hi_, axis=1)
        return self.scaler_.transform(Xc[self.keep_].values).astype(np.float32)

    def fit_transform(self, X):
        return self.fit(X).transform(X)


SPLITS = {}
for key, d in COHORTS.items():
    Xtr, Xho, ytr, yho = train_test_split(
        d["X"], d["y"], test_size=CFG["HOLDOUT_FRAC"],
        stratify=d["y"], random_state=CFG["SEED"])
    SPLITS[key] = dict(X_dev=Xtr.reset_index(drop=True), y_dev=ytr,
                       X_hold=Xho.reset_index(drop=True), y_hold=yho)
    print(f"{key}: development={Xtr.shape} {Counter(ytr)} | "
          f"HOLD-OUT={Xho.shape} {Counter(yho)}")
print("\nThe hold-out rows are the 'new data'. They are excluded from every "
      "fold, every scaler fit and every training loop until CELL 13.")


# %% ============  CELL 6 — TORCH DATASETS, PAIR & TRIPLET SAMPLERS  =========
banner("CELL 6 — PAIR / TRIPLET SAMPLING MACHINERY")


class TabularDS(Dataset):
    def __init__(self, X, y):
        self.X = torch.tensor(np.asarray(X), dtype=torch.float32)
        self.y = torch.tensor(np.asarray(y), dtype=torch.long)

    def __len__(self):
        return len(self.y)

    def __getitem__(self, i):
        return self.X[i], self.y[i]


class PairDataset(Dataset):
    """Balanced positive/negative pairs for the SIAMESE contrastive loss.
    Pairs are re-drawn every epoch (call .resample()) so the network sees far
    more of the O(n^2) pair space than a fixed list would give."""

    def __init__(self, X, y, pairs_per_epoch=None, rng=None):
        self.X = torch.tensor(np.asarray(X), dtype=torch.float32)
        self.y = np.asarray(y)
        self.idx_by_cls = {c: np.where(self.y == c)[0] for c in np.unique(self.y)}
        self.n_pairs = pairs_per_epoch or 4 * len(self.y)
        self.rng = rng or np.random.default_rng(CFG["SEED"])
        self.resample()

    def resample(self):
        a, b, sim = [], [], []
        classes = list(self.idx_by_cls)
        half = self.n_pairs // 2
        for _ in range(half):                       # positive pairs
            c = classes[self.rng.integers(len(classes))]
            pool = self.idx_by_cls[c]
            if len(pool) < 2:
                continue
            i, j = self.rng.choice(pool, 2, replace=False)
            a.append(i); b.append(j); sim.append(1.0)
        for _ in range(half):                       # negative pairs
            c1, c2 = self.rng.choice(classes, 2, replace=False)
            a.append(self.rng.choice(self.idx_by_cls[c1]))
            b.append(self.rng.choice(self.idx_by_cls[c2]))
            sim.append(0.0)
        self.a = np.array(a); self.b = np.array(b)
        self.sim = np.array(sim, dtype=np.float32)

    def __len__(self):
        return len(self.sim)

    def __getitem__(self, i):
        return self.X[self.a[i]], self.X[self.b[i]], torch.tensor(self.sim[i])


class BalancedBatchSampler(torch.utils.data.Sampler):
    """Yields batches containing K samples from each of P classes. Required for
    the TRIPLET network: batch-hard mining is meaningless if a batch happens to
    contain a single class, which is very likely on an imbalanced cohort."""

    def __init__(self, y, n_classes=2, n_per_class=16, n_batches=None):
        self.y = np.asarray(y)
        self.idx_by_cls = {c: np.where(self.y == c)[0] for c in np.unique(self.y)}
        self.n_classes = min(n_classes, len(self.idx_by_cls))
        self.n_per_class = n_per_class
        self.n_batches = n_batches or max(1, len(y) // (self.n_classes * n_per_class))
        self.rng = np.random.default_rng(CFG["SEED"])

    def __iter__(self):
        for _ in range(self.n_batches):
            batch = []
            for c in self.rng.choice(list(self.idx_by_cls), self.n_classes,
                                     replace=False):
                pool = self.idx_by_cls[c]
                batch += list(self.rng.choice(
                    pool, self.n_per_class, replace=len(pool) < self.n_per_class))
            self.rng.shuffle(batch)
            yield batch

    def __len__(self):
        return self.n_batches


print("PairDataset (contrastive) and BalancedBatchSampler (batch-hard triplets) ready.")


# %% ================  CELL 7 — MODEL ARCHITECTURES (PyTorch)  ===============
banner("CELL 7 — MODEL DEFINITIONS")


# ------------------------------- shared encoder -----------------------------
class TabularEncoder(nn.Module):
    """MLP encoder f: R^F -> R^d, output L2-normalised so that Euclidean
    distance on the unit sphere is bounded in [0, 2] and the contrastive /
    triplet margins have a consistent scale."""

    def __init__(self, in_dim, emb_dim=CFG["EMB_DIM"], hidden=CFG["HIDDEN"],
                 p=CFG["DROPOUT"]):
        super().__init__()
        layers, prev = [], in_dim
        for h in hidden:
            layers += [nn.Linear(prev, h), nn.BatchNorm1d(h), nn.ReLU(),
                       nn.Dropout(p)]
            prev = h
        layers += [nn.Linear(prev, emb_dim)]
        self.net = nn.Sequential(*layers)

    def forward(self, x):
        return F.normalize(self.net(x), p=2, dim=1)


# ------------------------------- 1. SIAMESE ---------------------------------
class SiameseNetwork(nn.Module):
    """Two identical towers with TIED weights. Trained on pairs: same-class
    pairs are pulled together, different-class pairs pushed apart beyond a
    margin. Classification is done afterwards in the learned embedding space."""

    def __init__(self, in_dim, emb_dim=CFG["EMB_DIM"]):
        super().__init__()
        self.encoder = TabularEncoder(in_dim, emb_dim)

    def forward_once(self, x):
        return self.encoder(x)

    def forward(self, x1, x2):
        return self.encoder(x1), self.encoder(x2)


class ContrastiveLoss(nn.Module):
    """L = y*d^2 + (1-y)*max(0, m-d)^2  with y=1 meaning 'same class'."""

    def __init__(self, margin=CFG["MARGIN_CONTRAST"]):
        super().__init__()
        self.m = margin

    def forward(self, e1, e2, sim):
        d = F.pairwise_distance(e1, e2, p=2)
        return torch.mean(sim * d.pow(2) +
                          (1 - sim) * F.relu(self.m - d).pow(2))


# ------------------------------- 2. TRIPLET ---------------------------------
class TripletNetwork(nn.Module):
    """Three towers with tied weights (anchor / positive / negative). Same
    encoder as the Siamese model — the difference is the LOSS: a triplet loss
    optimises a *relative* ordering d(a,p) + margin < d(a,n) rather than the
    absolute distances the contrastive loss fixes."""

    def __init__(self, in_dim, emb_dim=CFG["EMB_DIM"]):
        super().__init__()
        self.encoder = TabularEncoder(in_dim, emb_dim)

    def forward_once(self, x):
        return self.encoder(x)

    def forward(self, a, p, n):
        return self.encoder(a), self.encoder(p), self.encoder(n)


class BatchHardTripletLoss(nn.Module):
    """Online batch-hard mining: for every anchor take the FARTHEST positive
    and the NEAREST negative inside the batch. Random triplets go slack after a
    few epochs; batch-hard keeps supplying informative gradient."""

    def __init__(self, margin=CFG["MARGIN_TRIPLET"]):
        super().__init__()
        self.m = margin

    def forward(self, emb, labels):
        d = torch.cdist(emb, emb, p=2)
        lab = labels.unsqueeze(0)
        pos_mask = (lab == lab.t()).float() - torch.eye(len(labels),
                                                        device=emb.device)
        neg_mask = (lab != lab.t()).float()
        hardest_pos = (d * pos_mask).max(dim=1)[0]
        d_neg = d + (1 - neg_mask) * 1e6
        hardest_neg = d_neg.min(dim=1)[0]
        valid = (pos_mask.sum(1) > 0) & (neg_mask.sum(1) > 0)
        if valid.sum() == 0:
            return emb.sum() * 0.0
        loss = F.relu(hardest_pos - hardest_neg + self.m)[valid]
        return loss.mean()


# ------------------------------- 3. CAPSNET ---------------------------------
def squash(s, dim=-1, eps=1e-8):
    """Non-linearity that shrinks vector LENGTH into [0,1) while preserving
    orientation — the length is then read as an existence probability."""
    n2 = (s ** 2).sum(dim=dim, keepdim=True)
    return (n2 / (1.0 + n2)) * s / torch.sqrt(n2 + eps)


class PrimaryCaps(nn.Module):
    def __init__(self, in_ch, out_ch, caps_dim, kernel=3, stride=2):
        super().__init__()
        self.caps_dim = caps_dim
        self.conv = nn.Conv1d(in_ch, out_ch * caps_dim, kernel_size=kernel,
                              stride=stride, padding=1)

    def forward(self, x):
        u = self.conv(x)                                    # B, C*d, L
        B = u.size(0)
        u = u.view(B, -1, self.caps_dim, u.size(-1))         # B, C, d, L
        u = u.permute(0, 1, 3, 2).reshape(B, -1, self.caps_dim)
        return squash(u)                                     # B, N_caps, d


class DigitCaps(nn.Module):
    """Class capsules produced by DYNAMIC ROUTING-BY-AGREEMENT."""

    def __init__(self, n_in, d_in, n_out, d_out, iters=CFG["ROUTING_ITERS"]):
        super().__init__()
        self.n_out, self.d_out, self.iters = n_out, d_out, iters
        self.W = nn.Parameter(0.01 * torch.randn(n_in, n_out, d_out, d_in))

    def forward(self, u):
        B = u.size(0)
        # u: (B, n_in, d_in)  W: (n_in, n_out, d_out, d_in) -> (B,n_in,n_out,d_out)
        u_hat = torch.einsum("nodi,bni->bnod", self.W, u)
        u_hat_d = u_hat.detach()
        b = torch.zeros(B, u.size(1), self.n_out, device=u.device)
        for it in range(self.iters):
            c = F.softmax(b, dim=2).unsqueeze(-1)
            src = u_hat if it == self.iters - 1 else u_hat_d
            v = squash((c * src).sum(dim=1))                 # B, n_out, d_out
            if it < self.iters - 1:
                b = b + (u_hat_d * v.unsqueeze(1)).sum(-1)
        return v


class CapsuleNetwork(nn.Module):
    """CapsNet adapted from images to TABULAR data: the feature vector is
    treated as a length-F 1-D signal, Conv1d extracts local feature groups,
    those become primary capsules, and routing-by-agreement produces one
    capsule per class. ||v_c|| is the probability of class c. A reconstruction
    decoder regularises the capsule representation."""

    def __init__(self, in_dim, n_classes=2, conv_ch=128,
                 prim_caps=16, prim_dim=CFG["CAPS_PRIM_DIM"],
                 out_dim=CFG["CAPS_OUT_DIM"]):
        super().__init__()
        self.in_dim, self.n_classes, self.out_dim = in_dim, n_classes, out_dim
        self.conv = nn.Sequential(
            nn.Conv1d(1, conv_ch, kernel_size=3, padding=1),
            nn.BatchNorm1d(conv_ch), nn.ReLU(),
        )
        self.primary = PrimaryCaps(conv_ch, prim_caps, prim_dim)
        with torch.no_grad():
            n_caps = self.primary(self.conv(torch.zeros(2, 1, in_dim))).shape[1]
        self.n_caps = n_caps
        self.digit = DigitCaps(n_caps, prim_dim, n_classes, out_dim)
        self.decoder = nn.Sequential(
            nn.Linear(out_dim * n_classes, 128), nn.ReLU(),
            nn.Linear(128, 128), nn.ReLU(),
            nn.Linear(128, in_dim),
        )

    def forward(self, x, y_onehot=None):
        h = self.conv(x.unsqueeze(1))
        v = self.digit(self.primary(h))                      # B, n_classes, d
        lengths = v.norm(dim=-1)                             # B, n_classes
        if y_onehot is None:
            y_onehot = F.one_hot(lengths.argmax(1),
                                 self.n_classes).float()
        recon = self.decoder((v * y_onehot.unsqueeze(-1)).flatten(1))
        return lengths, recon, v


class MarginLoss(nn.Module):
    """Capsule margin loss, with per-class weights so the rare PCOS-positive
    capsule is not drowned out on an imbalanced cohort."""

    def __init__(self, m_pos=0.9, m_neg=0.1, lam=0.5, class_weight=None):
        super().__init__()
        self.mp, self.mn, self.lam = m_pos, m_neg, lam
        self.cw = class_weight

    def forward(self, lengths, y_onehot):
        L = (y_onehot * F.relu(self.mp - lengths).pow(2) +
             self.lam * (1 - y_onehot) * F.relu(lengths - self.mn).pow(2))
        if self.cw is not None:
            L = L * self.cw.to(lengths.device).unsqueeze(0)
        return L.sum(dim=1).mean()


print("Defined: SiameseNetwork + ContrastiveLoss")
print("         TripletNetwork + BatchHardTripletLoss")
print("         CapsuleNetwork (PrimaryCaps -> dynamic routing -> DigitCaps) "
      "+ MarginLoss + reconstruction decoder")


# %% ====================  CELL 8 — TRAINING ROUTINES  =======================
banner("CELL 8 — TRAINING ROUTINES")


def class_weights(y):
    cnt = np.bincount(y, minlength=2).astype(float)
    w = cnt.sum() / (2.0 * np.maximum(cnt, 1))
    return torch.tensor(w, dtype=torch.float32)



def inner_split(y, frac=0.18, min_pos=2, seed=CFG["SEED"]):
    """Carve a small stratified inner-validation set out of a TRAINING fold so
    training can early-stop honestly. Returns (train_idx, val_idx) or
    (all_idx, None) when the fold is too small / too imbalanced to spare rows."""
    idx = np.arange(len(y))
    counts = np.bincount(y, minlength=2)
    if counts.min() < 2 * min_pos or len(y) < 40:
        return idx, None
    tr, va = train_test_split(idx, test_size=frac, stratify=y,
                              random_state=seed)
    if np.bincount(y[va], minlength=2).min() < min_pos:
        return idx, None
    return tr, va


def fit_threshold(y_true, prob, objective="balanced_accuracy"):
    """Choose the decision cut-off on data the model did NOT train on (the
    inner-validation split of the training fold). A fixed 0.5 is arbitrary for
    an imbalanced cohort — CapsNet capsule-length ratios in particular rarely
    cross 0.5 for the rare class, which silently turns the model into an
    all-negative predictor with a perfectly good AUC."""
    if prob is None or len(np.unique(y_true)) < 2:
        return CFG["THRESHOLD"]
    grid = np.linspace(0.05, 0.95, 91)
    fn = (balanced_accuracy_score if objective == "balanced_accuracy"
          else (lambda a, b: f1_score(a, b, zero_division=0)))
    scores = [fn(y_true, (prob >= t).astype(int)) for t in grid]
    return float(grid[int(np.argmax(scores))])


def _inner_score_metric(model, Xtr, ytr, Xva, yva):
    etr, eva = embed(model, Xtr), embed(model, Xva)
    if len(np.unique(yva)) < 2:
        return np.nan
    p = EmbeddingClassifier().fit(etr, ytr).predict_proba(eva)[:, 1]
    return roc_auc_score(yva, p)


# ---- shared head: turn an embedding space into calibrated class scores ------
class EmbeddingClassifier:
    """Both metric-learning models produce an EMBEDDING, not a probability.
    This head converts one into the other:
      * hard label  = nearest class centroid
      * probability = softmax(-distance / T) over the two centroids
      * kNN vote reported alongside as an independent sanity check
    Everything is fitted on TRAIN embeddings only."""

    def __init__(self, temp=CFG["SOFTMAX_TEMP"], k=5):
        self.temp, self.k = temp, k

    def fit(self, emb_tr, y_tr):
        self.classes_ = np.array([0, 1])
        self.centroids_ = np.stack([
            emb_tr[y_tr == c].mean(0) if (y_tr == c).sum() else emb_tr.mean(0)
            for c in self.classes_])
        self.centroids_ /= (np.linalg.norm(self.centroids_, axis=1,
                                           keepdims=True) + 1e-12)
        k = int(min(self.k, max(1, (y_tr == 1).sum(), 1)))
        self.knn_ = KNeighborsClassifier(n_neighbors=max(1, min(k, len(y_tr) - 1)),
                                         weights="distance").fit(emb_tr, y_tr)
        return self

    def _dist(self, emb):
        return np.linalg.norm(emb[:, None, :] - self.centroids_[None], axis=2)

    def predict_proba(self, emb):
        logits = -self._dist(emb) / self.temp
        logits -= logits.max(1, keepdims=True)
        e = np.exp(logits)
        return e / e.sum(1, keepdims=True)

    def predict(self, emb, thr=CFG["THRESHOLD"]):
        return (self.predict_proba(emb)[:, 1] >= thr).astype(int)

    def predict_knn(self, emb):
        return self.knn_.predict(emb)


# ------------------------------ 1. SIAMESE ----------------------------------
def train_siamese(Xtr, ytr, epochs=None, verbose=False, patience=8, eval_every=5):
    """Early stopping on an inner validation split: cohort A tolerates the full
    schedule, cohort B (n~137) starts memorising after ~30 epochs. Letting the
    data decide beats hand-picking an epoch count per cohort."""
    epochs = epochs or CFG["EPOCHS_METRIC"]
    set_seed()
    tr_i, va_i = inner_split(np.asarray(ytr))
    Xin, yin = Xtr[tr_i], np.asarray(ytr)[tr_i]
    Xva, yva = (Xtr[va_i], np.asarray(ytr)[va_i]) if va_i is not None else (None, None)
    Xtr, ytr = Xin, yin
    best_score, best_state, bad = -np.inf, None, 0
    model = SiameseNetwork(Xtr.shape[1]).to(DEVICE)
    crit = ContrastiveLoss()
    opt = torch.optim.AdamW(model.parameters(), lr=CFG["LR"],
                            weight_decay=CFG["WEIGHT_DECAY"])
    sched = torch.optim.lr_scheduler.CosineAnnealingLR(opt, T_max=epochs)
    ds = PairDataset(Xtr, ytr)
    dl = DataLoader(ds, batch_size=CFG["BATCH"], shuffle=True, drop_last=True)
    hist = []
    for ep in range(epochs):
        ds.resample()
        model.train(); tot = n = 0
        for x1, x2, sim in dl:
            x1, x2, sim = x1.to(DEVICE), x2.to(DEVICE), sim.to(DEVICE)
            opt.zero_grad()
            e1, e2 = model(x1, x2)
            loss = crit(e1, e2, sim)
            loss.backward(); opt.step()
            tot += loss.item() * len(sim); n += len(sim)
        sched.step(); hist.append(tot / max(n, 1))
        if Xva is not None and (ep + 1) % eval_every == 0:
            sc = _inner_score_metric(model, Xtr, ytr, Xva, yva)
            if np.isfinite(sc) and sc > best_score + 1e-4:
                best_score, bad = sc, 0
                best_state = {k: v.detach().clone()
                              for k, v in model.state_dict().items()}
            else:
                bad += 1
                if bad >= patience:
                    if verbose:
                        print(f"    siamese early stop @ep{ep+1} "
                              f"(inner AUC={best_score:.3f})")
                    break
        if verbose and (ep + 1) % 30 == 0:
            print(f"    siamese ep{ep+1:3d} contrastive={hist[-1]:.4f}")
    if best_state is not None:
        model.load_state_dict(best_state)
    return model, hist


# ------------------------------ 2. TRIPLET ----------------------------------
def train_triplet(Xtr, ytr, epochs=None, verbose=False, patience=8, eval_every=5):
    epochs = epochs or CFG["EPOCHS_METRIC"]
    set_seed()
    tr_i, va_i = inner_split(np.asarray(ytr))
    Xva, yva = (Xtr[va_i], np.asarray(ytr)[va_i]) if va_i is not None else (None, None)
    Xtr, ytr = Xtr[tr_i], np.asarray(ytr)[tr_i]
    best_score, best_state, bad = -np.inf, None, 0
    model = TripletNetwork(Xtr.shape[1]).to(DEVICE)
    crit = BatchHardTripletLoss()
    opt = torch.optim.AdamW(model.parameters(), lr=CFG["LR"],
                            weight_decay=CFG["WEIGHT_DECAY"])
    sched = torch.optim.lr_scheduler.CosineAnnealingLR(opt, T_max=epochs)
    n_per = max(4, min(16, int(np.bincount(ytr, minlength=2).min())))
    ds = TabularDS(Xtr, ytr)
    sampler = BalancedBatchSampler(ytr, n_classes=2, n_per_class=n_per,
                                   n_batches=max(4, len(ytr) // (2 * n_per)))
    dl = DataLoader(ds, batch_sampler=sampler)
    hist = []
    for ep in range(epochs):
        model.train(); tot = n = 0
        for xb, yb in dl:
            xb, yb = xb.to(DEVICE), yb.to(DEVICE)
            opt.zero_grad()
            emb = model.forward_once(xb)
            loss = crit(emb, yb)
            if loss.requires_grad:
                loss.backward(); opt.step()
            tot += loss.item() * len(yb); n += len(yb)
        sched.step(); hist.append(tot / max(n, 1))
        if Xva is not None and (ep + 1) % eval_every == 0:
            sc = _inner_score_metric(model, Xtr, ytr, Xva, yva)
            if np.isfinite(sc) and sc > best_score + 1e-4:
                best_score, bad = sc, 0
                best_state = {k: v.detach().clone()
                              for k, v in model.state_dict().items()}
            else:
                bad += 1
                if bad >= patience:
                    if verbose:
                        print(f"    triplet early stop @ep{ep+1} "
                              f"(inner AUC={best_score:.3f})")
                    break
        if verbose and (ep + 1) % 30 == 0:
            print(f"    triplet ep{ep+1:3d} batch-hard={hist[-1]:.4f}")
    if best_state is not None:
        model.load_state_dict(best_state)
    return model, hist


@torch.no_grad()
def embed(model, X, bs=512):
    model.eval()
    out = []
    for i in range(0, len(X), bs):
        xb = torch.tensor(np.asarray(X[i:i + bs]), dtype=torch.float32).to(DEVICE)
        out.append(model.forward_once(xb).cpu().numpy())
    return np.vstack(out)


# ------------------------------ 3. CAPSNET ----------------------------------
def train_capsnet(Xtr, ytr, epochs=None, verbose=False, patience=8, eval_every=5):
    epochs = epochs or CFG["EPOCHS_CAPS"]
    set_seed()
    tr_i, va_i = inner_split(np.asarray(ytr))
    Xva, yva = (Xtr[va_i], np.asarray(ytr)[va_i]) if va_i is not None else (None, None)
    Xtr, ytr = Xtr[tr_i], np.asarray(ytr)[tr_i]
    best_score, best_state, bad = -np.inf, None, 0
    model = CapsuleNetwork(Xtr.shape[1]).to(DEVICE)
    crit = MarginLoss(class_weight=class_weights(ytr))
    opt = torch.optim.AdamW(model.parameters(), lr=CFG["LR"],
                            weight_decay=CFG["WEIGHT_DECAY"])
    sched = torch.optim.lr_scheduler.CosineAnnealingLR(opt, T_max=epochs)
    dl = DataLoader(TabularDS(Xtr, ytr), batch_size=CFG["BATCH"], shuffle=True)
    hist, hist_m, hist_r = [], [], []
    for ep in range(epochs):
        model.train(); tot = tm = tr = n = 0
        for xb, yb in dl:
            xb, yb = xb.to(DEVICE), yb.to(DEVICE)
            oh = F.one_hot(yb, 2).float()
            opt.zero_grad()
            lengths, recon, _ = model(xb, oh)
            m_loss = crit(lengths, oh)
            r_loss = F.mse_loss(recon, xb, reduction="mean")
            loss = m_loss + CFG["RECON_WEIGHT"] * r_loss * xb.shape[1]
            loss.backward(); opt.step()
            tot += loss.item() * len(yb); tm += m_loss.item() * len(yb)
            tr += r_loss.item() * len(yb); n += len(yb)
        sched.step()
        hist.append(tot / n); hist_m.append(tm / n); hist_r.append(tr / n)
        if Xva is not None and (ep + 1) % eval_every == 0:
            pv, _ = capsnet_predict(model, Xva)
            sc = (roc_auc_score(yva, pv[:, 1])
                  if len(np.unique(yva)) > 1 else np.nan)
            if np.isfinite(sc) and sc > best_score + 1e-4:
                best_score, bad = sc, 0
                best_state = {k: v.detach().clone()
                              for k, v in model.state_dict().items()}
            else:
                bad += 1
                if bad >= patience:
                    if verbose:
                        print(f"    capsnet early stop @ep{ep+1} "
                              f"(inner AUC={best_score:.3f})")
                    break
        if verbose and (ep + 1) % 30 == 0:
            print(f"    capsnet ep{ep+1:3d} total={hist[-1]:.4f} "
                  f"margin={hist_m[-1]:.4f} recon={hist_r[-1]:.4f}")
    if best_state is not None:
        model.load_state_dict(best_state)
    return model, dict(total=hist, margin=hist_m, recon=hist_r)


@torch.no_grad()
def capsnet_predict(model, X, bs=512):
    """Returns (probabilities, capsule lengths). ||v_c|| is the capsule's own
    existence probability; normalising the two lengths gives a proper score."""
    model.eval()
    P, L = [], []
    for i in range(0, len(X), bs):
        xb = torch.tensor(np.asarray(X[i:i + bs]), dtype=torch.float32).to(DEVICE)
        lengths, _, _ = model(xb)
        l = lengths.cpu().numpy()
        L.append(l)
        P.append(l / np.clip(l.sum(1, keepdims=True), 1e-9, None))
    return np.vstack(P), np.vstack(L)


print("Trainers ready: train_siamese / train_triplet / train_capsnet")


# %% =====================  CELL 9 — EVALUATION METRICS  =====================
banner("CELL 9 — METRIC SUITE")


def full_metrics(y_true, y_prob, thr=CFG["THRESHOLD"], y_pred=None):
    """Every metric that is meaningful for imbalanced binary clinical
    classification. Accuracy alone is misleading here: predicting 'No PCOS'
    for everyone already scores 0.78 on cohort B."""
    y_true = np.asarray(y_true).astype(int)
    y_prob = np.asarray(y_prob, dtype=float)
    if y_pred is None:
        y_pred = (y_prob >= thr).astype(int)
    y_pred = np.asarray(y_pred).astype(int)

    cm = confusion_matrix(y_true, y_pred, labels=[0, 1])
    tn, fp, fn, tp = cm.ravel()
    spec = tn / (tn + fp) if (tn + fp) else np.nan
    npv = tn / (tn + fn) if (tn + fn) else np.nan
    ppv = tp / (tp + fp) if (tp + fp) else np.nan
    sens = tp / (tp + fn) if (tp + fn) else np.nan
    both = len(np.unique(y_true)) > 1

    return dict(
        Accuracy          = accuracy_score(y_true, y_pred),
        Balanced_Accuracy = balanced_accuracy_score(y_true, y_pred),
        Precision_PPV     = ppv,
        Recall_Sensitivity= sens,
        Specificity       = spec,
        NPV               = npv,
        F1                = f1_score(y_true, y_pred, zero_division=0),
        F2                = fbeta_score(y_true, y_pred, beta=2, zero_division=0),
        MCC               = matthews_corrcoef(y_true, y_pred) if both else np.nan,
        Cohen_Kappa       = cohen_kappa_score(y_true, y_pred) if both else np.nan,
        ROC_AUC           = roc_auc_score(y_true, y_prob) if both else np.nan,
        PR_AUC            = average_precision_score(y_true, y_prob) if both else np.nan,
        LogLoss           = log_loss(y_true, np.clip(y_prob, 1e-7, 1 - 1e-7),
                                     labels=[0, 1]),
        Brier             = brier_score_loss(y_true, np.clip(y_prob, 0, 1)),
        Youden_J          = (sens + spec - 1) if both else np.nan,
        TN=int(tn), FP=int(fp), FN=int(fn), TP=int(tp),
    )


def summarise_folds(fold_rows):
    """mean ± std across folds for every metric."""
    df = pd.DataFrame(fold_rows)
    num = df.select_dtypes(include=[np.number])
    return pd.DataFrame({"mean": num.mean(), "std": num.std(ddof=1)})


def best_threshold(y_true, y_prob):
    """Youden-J optimal cut-off — reported as a supplement, never used to
    retro-fit the headline numbers."""
    fpr, tpr, thr = roc_curve(y_true, y_prob)
    j = tpr - fpr
    return float(thr[int(np.argmax(j))]), float(j.max())


METRIC_ORDER = ["Accuracy", "Balanced_Accuracy", "Precision_PPV",
                "Recall_Sensitivity", "Specificity", "NPV", "F1", "F2", "MCC",
                "Cohen_Kappa", "ROC_AUC", "PR_AUC", "LogLoss", "Brier", "Youden_J"]
print("Metrics per model: " + ", ".join(METRIC_ORDER))
print("Plus raw confusion counts TN/FP/FN/TP and embedding silhouette.")


# %% ==============  CELL 10 — STRATIFIED 5-FOLD CROSS-VALIDATION  ===========
banner("CELL 10 — STRATIFIED 5-FOLD CROSS-VALIDATION")


def run_cv(X_dev, y_dev, cohort_tag, n_folds=CFG["N_FOLDS"], verbose=True):
    """One StratifiedKFold loop shared by all three models, so the folds are
    identical and the comparison is paired. The preprocessor is re-fitted
    inside every fold — nothing about the validation rows leaks into scaling
    or imputation."""
    skf = StratifiedKFold(n_splits=n_folds, shuffle=True,
                          random_state=CFG["SEED"])
    models = ["Siamese", "Triplet", "CapsNet"]
    fold_metrics = {m: [] for m in models}
    oof_prob = {m: np.zeros(len(y_dev)) for m in models}
    oof_pred = {m: np.zeros(len(y_dev), dtype=int) for m in models}
    knn_acc = {m: [] for m in models if m != "CapsNet"}
    loss_hist = {m: [] for m in models}
    embeds = {m: np.zeros((len(y_dev), CFG["EMB_DIM"])) for m in
              ["Siamese", "Triplet"]}
    caps_lengths = np.zeros((len(y_dev), 2))
    sil = {m: [] for m in ["Siamese", "Triplet"]}
    t0 = time.time()

    thresholds = {m: [] for m in models}

    for fold, (tr, va) in enumerate(skf.split(X_dev, y_dev), 1):
        pre = Preprocessor().fit(X_dev.iloc[tr])
        Xtr, Xva = pre.transform(X_dev.iloc[tr]), pre.transform(X_dev.iloc[va])
        ytr, yva = y_dev[tr], y_dev[va]
        # the same inner split the trainers early-stop on — reused here to pick
        # the operating threshold, so the outer validation stays untouched
        in_tr, in_va = inner_split(ytr)

        # ---- Siamese -------------------------------------------------------
        sm, sh = train_siamese(Xtr, ytr)
        etr, eva = embed(sm, Xtr), embed(sm, Xva)
        thr_s = CFG["THRESHOLD"]
        if in_va is not None:
            h_in = EmbeddingClassifier().fit(etr[in_tr], ytr[in_tr])
            thr_s = fit_threshold(ytr[in_va],
                                  h_in.predict_proba(etr[in_va])[:, 1])
        thresholds["Siamese"].append(thr_s)
        head = EmbeddingClassifier().fit(etr, ytr)
        p = head.predict_proba(eva)[:, 1]
        oof_prob["Siamese"][va] = p
        oof_pred["Siamese"][va] = (p >= thr_s).astype(int)
        embeds["Siamese"][va] = eva
        fold_metrics["Siamese"].append(
            {**full_metrics(yva, p, thr=thr_s), "fold": fold, "threshold": thr_s})
        knn_acc["Siamese"].append(accuracy_score(yva, head.predict_knn(eva)))
        loss_hist["Siamese"].append(sh)
        sil["Siamese"].append(silhouette_score(eva, yva)
                              if len(np.unique(yva)) > 1 else np.nan)

        # ---- Triplet -------------------------------------------------------
        tm, th = train_triplet(Xtr, ytr)
        etr, eva = embed(tm, Xtr), embed(tm, Xva)
        thr_t = CFG["THRESHOLD"]
        if in_va is not None:
            h_in = EmbeddingClassifier().fit(etr[in_tr], ytr[in_tr])
            thr_t = fit_threshold(ytr[in_va],
                                  h_in.predict_proba(etr[in_va])[:, 1])
        thresholds["Triplet"].append(thr_t)
        head = EmbeddingClassifier().fit(etr, ytr)
        p = head.predict_proba(eva)[:, 1]
        oof_prob["Triplet"][va] = p
        oof_pred["Triplet"][va] = (p >= thr_t).astype(int)
        embeds["Triplet"][va] = eva
        fold_metrics["Triplet"].append(
            {**full_metrics(yva, p, thr=thr_t), "fold": fold, "threshold": thr_t})
        knn_acc["Triplet"].append(accuracy_score(yva, head.predict_knn(eva)))
        loss_hist["Triplet"].append(th)
        sil["Triplet"].append(silhouette_score(eva, yva)
                              if len(np.unique(yva)) > 1 else np.nan)

        # ---- CapsNet -------------------------------------------------------
        cm_, ch = train_capsnet(Xtr, ytr)
        thr_c = CFG["THRESHOLD"]
        if in_va is not None:
            pv_in, _ = capsnet_predict(cm_, Xtr[in_va])
            thr_c = fit_threshold(ytr[in_va], pv_in[:, 1])
        thresholds["CapsNet"].append(thr_c)
        pv, lv = capsnet_predict(cm_, Xva)
        p = pv[:, 1]
        oof_prob["CapsNet"][va] = p
        oof_pred["CapsNet"][va] = (p >= thr_c).astype(int)
        caps_lengths[va] = lv
        fold_metrics["CapsNet"].append(
            {**full_metrics(yva, p, thr=thr_c), "fold": fold, "threshold": thr_c})
        loss_hist["CapsNet"].append(ch["total"])

        if verbose:
            print(f"  [{cohort_tag}] fold {fold}/{n_folds} "
                  f"(n_tr={len(tr)}, n_va={len(va)})  " +
                  "  ".join(f"{m}: BA={fold_metrics[m][-1]['Balanced_Accuracy']:.3f} "
                            f"AUC={fold_metrics[m][-1]['ROC_AUC']:.3f}"
                            for m in models))

    res = dict(
        cohort=cohort_tag, models=models, fold_metrics=fold_metrics,
        summary={m: summarise_folds(fold_metrics[m]) for m in models},
        oof_prob=oof_prob, oof_pred=oof_pred, y=y_dev, knn_acc=knn_acc,
        loss_hist=loss_hist, embeds=embeds, caps_lengths=caps_lengths,
        silhouette=sil, thresholds=thresholds, seconds=time.time() - t0,
    )
    print(f"  -> {cohort_tag} CV finished in {res['seconds']:.1f}s")
    print("     tuned operating thresholds per fold: " +
          "  ".join(f"{m}={np.round(v, 2).tolist()}" for m, v in thresholds.items()))
    return res


# %% ================  CELL 11 — RESULT VISUALISATION SUITE  =================
banner("CELL 11 — RESULT VISUALISATIONS")

MODEL_COLORS = {"Siamese": "#0072b2", "Triplet": "#d55e00", "CapsNet": "#009e73"}


def visualise_cv(res, label):
    tag = res["cohort"]
    y = res["y"]
    models = res["models"]

    # 1. training-loss curves, all folds, per model
    plt.figure(figsize=(13, 3.6))
    for i, m in enumerate(models, 1):
        plt.subplot(1, 3, i)
        for f, h in enumerate(res["loss_hist"][m], 1):
            plt.plot(h, lw=1.1, alpha=.85, label=f"fold {f}")
        plt.title(f"{m} — training loss"); plt.xlabel("epoch")
        plt.ylabel("contrastive" if m == "Siamese" else
                   "batch-hard triplet" if m == "Triplet" else "margin+recon")
        plt.legend(fontsize=6)
    plt.suptitle(f"{label} — convergence across the 5 folds", y=1.03)
    plt.tight_layout(); savefig(f"{tag}_loss_curves")

    # 2. out-of-fold confusion matrices
    plt.figure(figsize=(12.5, 3.8))
    for i, m in enumerate(models, 1):
        plt.subplot(1, 3, i)
        cm = confusion_matrix(y, res["oof_pred"][m], labels=[0, 1])
        sns.heatmap(cm, annot=True, fmt="d", cmap="Blues", cbar=False,
                    xticklabels=["pred No", "pred Yes"],
                    yticklabels=["true No", "true Yes"])
        acc = accuracy_score(y, res["oof_pred"][m])
        ba = balanced_accuracy_score(y, res["oof_pred"][m])
        plt.title(f"{m}\nOOF acc={acc:.3f}  bal-acc={ba:.3f}")
    plt.suptitle(f"{label} — out-of-fold confusion matrices", y=1.05)
    plt.tight_layout(); savefig(f"{tag}_confusion_oof")

    # 3. normalised (row-wise) confusion matrices — sensitivity/specificity view
    plt.figure(figsize=(12.5, 3.8))
    for i, m in enumerate(models, 1):
        plt.subplot(1, 3, i)
        cm = confusion_matrix(y, res["oof_pred"][m], labels=[0, 1],
                              normalize="true")
        sns.heatmap(cm, annot=True, fmt=".3f", cmap="Purples", cbar=False,
                    vmin=0, vmax=1,
                    xticklabels=["pred No", "pred Yes"],
                    yticklabels=["true No", "true Yes"])
        plt.title(f"{m} (row-normalised)")
    plt.suptitle(f"{label} — specificity (top-left) / sensitivity (bottom-right)",
                 y=1.05)
    plt.tight_layout(); savefig(f"{tag}_confusion_normalised")

    # 4. ROC — per fold + OOF, per model
    plt.figure(figsize=(13, 4))
    for i, m in enumerate(models, 1):
        plt.subplot(1, 3, i)
        mean_fpr = np.linspace(0, 1, 100); tprs = []
        skf = StratifiedKFold(n_splits=CFG["N_FOLDS"], shuffle=True,
                              random_state=CFG["SEED"])
        for f, (_, va) in enumerate(skf.split(np.zeros(len(y)), y), 1):
            if len(np.unique(y[va])) < 2:
                continue
            fpr, tpr, _ = roc_curve(y[va], res["oof_prob"][m][va])
            tprs.append(np.interp(mean_fpr, fpr, tpr)); tprs[-1][0] = 0
            plt.plot(fpr, tpr, lw=.9, alpha=.4)
        fpr, tpr, _ = roc_curve(y, res["oof_prob"][m])
        auc_oof = roc_auc_score(y, res["oof_prob"][m])
        plt.plot(fpr, tpr, color=MODEL_COLORS[m], lw=2.4,
                 label=f"OOF AUC={auc_oof:.3f}")
        if tprs:
            mt = np.mean(tprs, axis=0)
            plt.plot(mean_fpr, mt, "k--", lw=1.4,
                     label=f"fold-mean AUC={_trapz(mt, mean_fpr):.3f}")
        plt.plot([0, 1], [0, 1], ":", c="grey")
        plt.xlabel("FPR (1 - specificity)"); plt.ylabel("TPR (sensitivity)")
        plt.title(m); plt.legend(fontsize=7, loc="lower right")
    plt.suptitle(f"{label} — ROC curves", y=1.04)
    plt.tight_layout(); savefig(f"{tag}_roc")

    # 5. Precision-Recall — the honest curve for an imbalanced positive class
    plt.figure(figsize=(5.6, 4.6))
    for m in models:
        pr, rc, _ = precision_recall_curve(y, res["oof_prob"][m])
        ap = average_precision_score(y, res["oof_prob"][m])
        plt.plot(rc, pr, color=MODEL_COLORS[m], lw=2, label=f"{m} AP={ap:.3f}")
    plt.axhline(y.mean(), ls=":", c="grey", label=f"baseline={y.mean():.3f}")
    plt.xlabel("recall"); plt.ylabel("precision"); plt.legend(fontsize=8)
    plt.title(f"{label}\nprecision–recall (out-of-fold)")
    savefig(f"{tag}_pr_curves")

    # 6. probability calibration + score histograms
    plt.figure(figsize=(12, 4))
    plt.subplot(1, 2, 1)
    for m in models:
        nb = min(8, max(3, len(y) // 25))
        pt, pp = calibration_curve(y, res["oof_prob"][m], n_bins=nb,
                                   strategy="quantile")
        plt.plot(pp, pt, "o-", color=MODEL_COLORS[m],
                 label=f"{m} Brier={brier_score_loss(y, res['oof_prob'][m]):.3f}")
    plt.plot([0, 1], [0, 1], ":", c="grey")
    plt.xlabel("mean predicted probability"); plt.ylabel("observed frequency")
    plt.title("calibration"); plt.legend(fontsize=7)
    plt.subplot(1, 2, 2)
    for m in models:
        sns.kdeplot(res["oof_prob"][m], label=m, color=MODEL_COLORS[m],
                    fill=True, alpha=.18, warn_singular=False)
    for m in models:
        plt.axvline(np.mean(res["thresholds"][m]), ls="--",
                    color=MODEL_COLORS[m], lw=1)
    plt.xlabel("P(PCOS)"); plt.title("predicted-score distribution "
                                     "(dashed = fitted thresholds)")
    plt.legend(fontsize=7)
    plt.suptitle(label, y=1.03); plt.tight_layout()
    savefig(f"{tag}_calibration")

    # 7. score separation by true class
    plt.figure(figsize=(12.5, 3.6))
    for i, m in enumerate(models, 1):
        plt.subplot(1, 3, i)
        dfp = pd.DataFrame({"p": res["oof_prob"][m],
                            "true": pd.Series(y).map({0: "No PCOS", 1: "PCOS"})})
        sns.violinplot(data=dfp, x="true", y="p", hue="true",
                       palette="Set2", legend=False, cut=0)
        sns.stripplot(data=dfp, x="true", y="p", color="k", size=2, alpha=.4)
        plt.axhline(np.mean(res["thresholds"][m]), ls="--", c="r", lw=1)
        plt.title(m); plt.xlabel(""); plt.ylabel("P(PCOS)")
    plt.suptitle(f"{label} — score separation by true class", y=1.05)
    plt.tight_layout(); savefig(f"{tag}_score_separation")

    # 8. learned embedding spaces (Siamese / Triplet) vs input space
    plt.figure(figsize=(12.5, 4))
    for i, m in enumerate(["Siamese", "Triplet"], 1):
        E = res["embeds"][m]
        per = max(5, min(30, (len(E) - 1) // 3))
        T = TSNE(n_components=2, perplexity=per, init="pca",
                 random_state=CFG["SEED"]).fit_transform(E)
        plt.subplot(1, 3, i)
        for cls, col, nm in [(0, "#00798c", "No PCOS"), (1, "#d1495b", "PCOS")]:
            plt.scatter(T[y == cls, 0], T[y == cls, 1], s=15, alpha=.75,
                        c=col, label=nm)
        s = silhouette_score(E, y) if len(np.unique(y)) > 1 else np.nan
        plt.title(f"{m} embedding\nsilhouette={s:.3f}"); plt.legend(fontsize=7)
    plt.subplot(1, 3, 3)
    C = res["caps_lengths"]
    for cls, col, nm in [(0, "#00798c", "No PCOS"), (1, "#d1495b", "PCOS")]:
        plt.scatter(C[y == cls, 0], C[y == cls, 1], s=15, alpha=.75, c=col,
                    label=nm)
    lim = [0, max(1.0, C.max() * 1.05)]
    plt.plot(lim, lim, ":", c="grey")
    plt.xlabel("||v_0||  (No-PCOS capsule)"); plt.ylabel("||v_1||  (PCOS capsule)")
    plt.title("CapsNet capsule lengths"); plt.legend(fontsize=7)
    plt.suptitle(f"{label} — what each model learned "
                 f"(input-space silhouette was {eda_results[tag]['input_silhouette']:.3f})"
                 if tag in eda_results else label, y=1.04)
    plt.tight_layout(); savefig(f"{tag}_embeddings")

    # 9. embedding distance distributions — the thing the metric losses optimise
    plt.figure(figsize=(11, 3.8))
    rng = np.random.default_rng(CFG["SEED"])
    for i, m in enumerate(["Siamese", "Triplet"], 1):
        E = res["embeds"][m]
        idx = rng.choice(len(E), min(len(E), 400), replace=False)
        D = np.linalg.norm(E[idx][:, None] - E[idx][None], axis=2)
        same = (y[idx][:, None] == y[idx][None])
        iu = np.triu_indices(len(idx), k=1)
        plt.subplot(1, 2, i)
        sns.kdeplot(D[iu][same[iu]], label="same class", fill=True, alpha=.3,
                    color="#00798c", warn_singular=False)
        sns.kdeplot(D[iu][~same[iu]], label="different class", fill=True,
                    alpha=.3, color="#d1495b", warn_singular=False)
        ov = (D[iu][same[iu]].mean(), D[iu][~same[iu]].mean())
        plt.title(f"{m}: mean d(same)={ov[0]:.3f}  d(diff)={ov[1]:.3f}")
        plt.xlabel("embedding distance"); plt.legend(fontsize=7)
    plt.suptitle(f"{label} — pairwise distance separation", y=1.04)
    plt.tight_layout(); savefig(f"{tag}_distance_dists")

    # 10. per-fold metric spread
    rows = []
    for m in models:
        for r in res["fold_metrics"][m]:
            for k in ["Balanced_Accuracy", "F1", "MCC", "ROC_AUC", "PR_AUC",
                      "Recall_Sensitivity", "Specificity"]:
                rows.append(dict(model=m, metric=k, value=r[k]))
    dfm = pd.DataFrame(rows)
    plt.figure(figsize=(12, 4.2))
    sns.boxplot(data=dfm, x="metric", y="value", hue="model",
                palette=MODEL_COLORS)
    sns.stripplot(data=dfm, x="metric", y="value", hue="model", dodge=True,
                  palette="dark:k", size=3, legend=False)
    plt.xticks(rotation=20, ha="right"); plt.ylim(0, 1.02)
    plt.title(f"{label} — per-fold metric spread (5 folds)")
    plt.legend(fontsize=8); savefig(f"{tag}_metric_boxplots")

    # 11. mean±std metric heatmap
    H = pd.DataFrame({m: res["summary"][m]["mean"] for m in models}).loc[METRIC_ORDER]
    S = pd.DataFrame({m: res["summary"][m]["std"] for m in models}).loc[METRIC_ORDER]
    ann = H.round(3).astype(str) + "\n±" + S.round(3).astype(str)
    plt.figure(figsize=(6.4, 7.4))
    sns.heatmap(H, annot=ann, fmt="", cmap="YlGnBu", cbar_kws=dict(shrink=.6),
                annot_kws=dict(fontsize=7.5))
    plt.title(f"{label}\n5-fold CV metrics (mean ± std)")
    savefig(f"{tag}_metric_heatmap")

    # 12. threshold sweep — where to actually set the operating point
    plt.figure(figsize=(13, 3.8))
    ths = np.linspace(0.05, 0.95, 91)
    for i, m in enumerate(models, 1):
        plt.subplot(1, 3, i)
        p = res["oof_prob"][m]
        for nm_, fn in [("bal-acc", balanced_accuracy_score),
                        ("F1", lambda a, b: f1_score(a, b, zero_division=0)),
                        ("MCC", matthews_corrcoef),
                        ("recall", lambda a, b: recall_score(a, b, zero_division=0)),
                        ("precision", lambda a, b: precision_score(a, b, zero_division=0))]:
            plt.plot(ths, [fn(y, (p >= t).astype(int)) for t in ths], lw=1.3,
                     label=nm_)
        bt, _ = best_threshold(y, p)
        plt.axvline(np.mean(res["thresholds"][m]), ls="--", c="k", lw=1)
        plt.axvline(bt, ls=":", c="r", lw=1.4)
        plt.title(f"{m}  (Youden-J opt thr={bt:.2f})"); plt.xlabel("threshold")
        plt.legend(fontsize=6)
    plt.suptitle(f"{label} — metric vs decision threshold "
                 f"(black = mean threshold actually fitted on inner validation)",
                 y=1.05)
    plt.tight_layout(); savefig(f"{tag}_threshold_sweep")

    print(f"[{tag}] result figures written.")


# %% ==========  CELL 12 — CLASSICAL BASELINES ON THE SAME FOLDS  ============
banner("CELL 12 — CLASSICAL BASELINES (same folds, paired comparison)")


def run_baselines(X_dev, y_dev, tag):
    """A deep model is only worth its complexity if it beats logistic
    regression. Same StratifiedKFold seed => the same 5 splits => a fair,
    paired comparison."""
    skf = StratifiedKFold(n_splits=CFG["N_FOLDS"], shuffle=True,
                          random_state=CFG["SEED"])
    builders = {
        "Majority (Dummy)": lambda: DummyClassifier(strategy="prior"),
        "LogisticRegression": lambda: LogisticRegression(
            max_iter=3000, class_weight="balanced", C=1.0),
        "RandomForest": lambda: RandomForestClassifier(
            n_estimators=400, class_weight="balanced_subsample",
            random_state=CFG["SEED"], n_jobs=-1),
    }
    fold_metrics = {k: [] for k in builders}
    oof = {k: np.zeros(len(y_dev)) for k in builders}
    for fold, (tr, va) in enumerate(skf.split(X_dev, y_dev), 1):
        pre = Preprocessor().fit(X_dev.iloc[tr])
        Xtr, Xva = pre.transform(X_dev.iloc[tr]), pre.transform(X_dev.iloc[va])
        for name, build in builders.items():
            clf = build().fit(Xtr, y_dev[tr])
            p = clf.predict_proba(Xva)[:, 1]
            oof[name][va] = p
            fold_metrics[name].append({**full_metrics(y_dev[va], p),
                                       "fold": fold})
    summary = {k: summarise_folds(v) for k, v in fold_metrics.items()}
    print(f"[{tag}] baselines (5-fold mean):")
    print(pd.DataFrame({k: summary[k]["mean"] for k in builders})
          .loc[["Accuracy", "Balanced_Accuracy", "F1", "MCC", "ROC_AUC",
                "PR_AUC"]].round(3).to_string())
    return dict(fold_metrics=fold_metrics, summary=summary, oof_prob=oof)


# %% ======  CELL 13 — FINAL FIT + PREDICTION ON THE UNSEEN HOLD-OUT  ========
banner("CELL 13 — FINAL MODELS + UNSEEN-DATA VERIFICATION")


def fit_final(X_dev, y_dev, tag, verbose=True):
    """Refit all three models on the FULL development set (no folds), and
    package everything needed to score brand-new patients: the fitted
    preprocessor, the three networks, and the two embedding heads."""
    pre = Preprocessor().fit(X_dev)
    Xd = pre.transform(X_dev)

    if verbose:
        print(f"[{tag}] final fit on n={len(y_dev)}, "
              f"{Xd.shape[1]} features kept (of {X_dev.shape[1]})")
    in_tr, in_va = inner_split(np.asarray(y_dev))
    thr = {}

    sm, sh = train_siamese(Xd, y_dev, verbose=verbose)
    e_all = embed(sm, Xd)
    thr["Siamese"] = (fit_threshold(
        y_dev[in_va], EmbeddingClassifier().fit(e_all[in_tr], y_dev[in_tr])
        .predict_proba(e_all[in_va])[:, 1]) if in_va is not None
        else CFG["THRESHOLD"])
    head_s = EmbeddingClassifier().fit(e_all, y_dev)

    tm, th = train_triplet(Xd, y_dev, verbose=verbose)
    e_all = embed(tm, Xd)
    thr["Triplet"] = (fit_threshold(
        y_dev[in_va], EmbeddingClassifier().fit(e_all[in_tr], y_dev[in_tr])
        .predict_proba(e_all[in_va])[:, 1]) if in_va is not None
        else CFG["THRESHOLD"])
    head_t = EmbeddingClassifier().fit(e_all, y_dev)

    cm_, ch = train_capsnet(Xd, y_dev, verbose=verbose)
    if in_va is not None:
        pv_in, _ = capsnet_predict(cm_, Xd[in_va])
        thr["CapsNet"] = fit_threshold(y_dev[in_va], pv_in[:, 1])
    else:
        thr["CapsNet"] = CFG["THRESHOLD"]
    thr["Ensemble"] = float(np.mean(list(thr.values())))
    print(f"[{tag}] operating thresholds fitted on inner validation: " +
          ", ".join(f"{k}={v:.2f}" for k, v in thr.items()))

    return dict(tag=tag, pre=pre, feature_names=pre.keep_, thresholds=thr,
                siamese=sm, head_s=head_s, triplet=tm, head_t=head_t,
                capsnet=cm_, loss=dict(Siamese=sh, Triplet=th, CapsNet=ch["total"]),
                y_dev=y_dev)


def predict_bundle(bundle, X_new: pd.DataFrame):
    """Score any raw dataframe with the same columns as the training frame.
    Returns per-model probabilities + a soft-voting ensemble."""
    Xn = bundle["pre"].transform(X_new)
    out = {}
    out["Siamese"] = bundle["head_s"].predict_proba(embed(bundle["siamese"], Xn))[:, 1]
    out["Triplet"] = bundle["head_t"].predict_proba(embed(bundle["triplet"], Xn))[:, 1]
    caps_p, caps_len = capsnet_predict(bundle["capsnet"], Xn)
    out["CapsNet"] = caps_p[:, 1]
    prob = pd.DataFrame(out)
    prob["Ensemble"] = prob[["Siamese", "Triplet", "CapsNet"]].mean(axis=1)
    thr = bundle.get("thresholds", {})
    pred = pd.DataFrame({m: (prob[m] >= thr.get(m, CFG["THRESHOLD"])).astype(int)
                         for m in prob.columns})
    return prob, pred, caps_len


def verify_holdout(bundle, X_hold, y_hold, label):
    """The real test: rows the pipeline has never seen in any form."""
    prob, pred, caps_len = predict_bundle(bundle, X_hold)
    tag = bundle["tag"]
    rows = {}
    for m in ["Siamese", "Triplet", "CapsNet", "Ensemble"]:
        rows[m] = full_metrics(y_hold, prob[m].values, y_pred=pred[m].values)
        rows[m]["Threshold"] = bundle.get("thresholds", {}).get(m, CFG["THRESHOLD"])
    table = pd.DataFrame(rows).loc[METRIC_ORDER + ["Threshold", "TN", "FP",
                                                   "FN", "TP"]]
    print(f"\n[{tag}] HOLD-OUT performance (n={len(y_hold)}, "
          f"positives={int(y_hold.sum())}) — never seen during training:")
    print(table.round(3).to_string())

    # confusion matrices on the hold-out
    plt.figure(figsize=(15, 3.6))
    for i, m in enumerate(["Siamese", "Triplet", "CapsNet", "Ensemble"], 1):
        plt.subplot(1, 4, i)
        sns.heatmap(confusion_matrix(y_hold, pred[m], labels=[0, 1]),
                    annot=True, fmt="d", cmap="Greens", cbar=False,
                    xticklabels=["pred No", "pred Yes"],
                    yticklabels=["true No", "true Yes"])
        plt.title(f"{m}\nbal-acc={rows[m]['Balanced_Accuracy']:.3f}")
    plt.suptitle(f"{label} — UNSEEN hold-out confusion matrices", y=1.05)
    plt.tight_layout(); savefig(f"{tag}_holdout_confusion")

    # ROC + PR on the hold-out
    plt.figure(figsize=(11, 4))
    plt.subplot(1, 2, 1)
    for m in ["Siamese", "Triplet", "CapsNet", "Ensemble"]:
        fpr, tpr, _ = roc_curve(y_hold, prob[m])
        plt.plot(fpr, tpr, lw=1.8,
                 label=f"{m} AUC={roc_auc_score(y_hold, prob[m]):.3f}")
    plt.plot([0, 1], [0, 1], ":", c="grey"); plt.legend(fontsize=8)
    plt.xlabel("FPR"); plt.ylabel("TPR"); plt.title("hold-out ROC")
    plt.subplot(1, 2, 2)
    for m in ["Siamese", "Triplet", "CapsNet", "Ensemble"]:
        pr, rc, _ = precision_recall_curve(y_hold, prob[m])
        plt.plot(rc, pr, lw=1.8,
                 label=f"{m} AP={average_precision_score(y_hold, prob[m]):.3f}")
    plt.axhline(y_hold.mean(), ls=":", c="grey")
    plt.xlabel("recall"); plt.ylabel("precision"); plt.title("hold-out PR")
    plt.legend(fontsize=8)
    plt.suptitle(label, y=1.03); plt.tight_layout()
    savefig(f"{tag}_holdout_roc_pr")

    # per-patient prediction table (first 20 unseen patients)
    show = prob.copy()
    show["true"] = y_hold
    show["ens_pred"] = pred["Ensemble"].values
    show["correct"] = (show["ens_pred"] == show["true"])
    print(f"\n[{tag}] first 15 unseen patients, predicted P(PCOS):")
    print(show.head(15).round(3).to_string())
    print(f"  ensemble agreement with ground truth on the hold-out: "
          f"{show['correct'].mean():.1%}")

    # model agreement matrix
    plt.figure(figsize=(4.6, 3.8))
    agree = pd.DataFrame(
        [[np.mean(pred[a] == pred[b]) for b in ["Siamese", "Triplet", "CapsNet"]]
         for a in ["Siamese", "Triplet", "CapsNet"]],
        index=["Siamese", "Triplet", "CapsNet"],
        columns=["Siamese", "Triplet", "CapsNet"])
    sns.heatmap(agree, annot=True, fmt=".3f", cmap="Oranges", vmin=0, vmax=1)
    plt.title(f"{label}\nhold-out inter-model agreement")
    savefig(f"{tag}_holdout_agreement")

    return dict(table=table, prob=prob, pred=pred, caps_len=caps_len,
                detail=show)


# %% =======  CELL 14 — FEATURE IMPORTANCE + NEW-PATIENT PREDICTION API  =====
banner("CELL 14 — PERMUTATION IMPORTANCE + NEW-PATIENT API")


def permutation_importance_bundle(bundle, X_hold, y_hold, label, n_repeats=5,
                                  top_k=20):
    """Model-agnostic importance: shuffle one column of the UNSEEN hold-out and
    measure the drop in ROC-AUC. Works identically for all three architectures
    because it only needs predict_bundle()."""
    rng = np.random.default_rng(CFG["SEED"])
    base_prob, _, _ = predict_bundle(bundle, X_hold)
    base = {m: roc_auc_score(y_hold, base_prob[m]) for m in
            ["Siamese", "Triplet", "CapsNet", "Ensemble"]}
    cols = list(X_hold.columns)
    imp = {m: np.zeros(len(cols)) for m in base}
    for j, c in enumerate(cols):
        for _ in range(n_repeats):
            Xp = X_hold.copy()
            Xp[c] = rng.permutation(Xp[c].values)
            pp, _, _ = predict_bundle(bundle, Xp)
            for m in base:
                imp[m][j] += (base[m] - roc_auc_score(y_hold, pp[m])) / n_repeats
    dfi = pd.DataFrame(imp, index=cols)
    top = dfi["Ensemble"].sort_values(ascending=False).head(top_k).index

    plt.figure(figsize=(9, max(4, .32 * len(top))))
    dfi.loc[top, ["Siamese", "Triplet", "CapsNet"]].iloc[::-1].plot(
        kind="barh", ax=plt.gca(), color=[MODEL_COLORS[m] for m in
                                          ["Siamese", "Triplet", "CapsNet"]])
    plt.axvline(0, c="k", lw=.8); plt.xlabel("drop in hold-out ROC-AUC when shuffled")
    plt.title(f"{label}\npermutation importance (top {len(top)})")
    plt.yticks(fontsize=7); plt.legend(fontsize=8)
    savefig(f"{bundle['tag']}_perm_importance")

    print(f"[{bundle['tag']}] top 10 drivers (ensemble):")
    print(dfi["Ensemble"].sort_values(ascending=False).head(10).round(4).to_string())
    return dfi


def predict_new_patients(bundle, records, template: pd.DataFrame, verbose=True):
    """Score brand-new patient(s) supplied as dict(s) of raw feature values.
    Anything you leave out is filled with the training median, so partial
    records are allowed — the preprocessor handles it."""
    if isinstance(records, dict):
        records = [records]
    frame = pd.DataFrame(records)
    unknown = [c for c in frame.columns if c not in template.columns]
    if unknown:
        raise KeyError(f"unknown feature name(s): {unknown}")
    new = pd.DataFrame(np.nan, index=range(len(frame)), columns=template.columns)
    for c in frame.columns:
        new[c] = frame[c].values
    prob, pred, caps_len = predict_bundle(bundle, new)
    prob = prob.round(4)
    prob["Ensemble_label"] = np.where(pred["Ensemble"].values == 1,
                                      "PCOS", "No PCOS")
    prob["capsule_len_No"] = caps_len[:, 0].round(3)
    prob["capsule_len_PCOS"] = caps_len[:, 1].round(3)
    if verbose:
        print(f"\n[{bundle['tag']}] prediction for {len(new)} NEW patient "
              f"record(s) ({len(frame.columns)}/{len(template.columns)} fields given):")
        print(prob.to_string())
    return prob


def save_bundle(bundle, outdir=OUTDIR):
    path = os.path.join(outdir, f"pcos_{bundle['tag']}_models.pt")
    torch.save(dict(
        siamese=bundle["siamese"].state_dict(),
        triplet=bundle["triplet"].state_dict(),
        capsnet=bundle["capsnet"].state_dict(),
        feature_names=bundle["feature_names"],
        centroids_s=bundle["head_s"].centroids_,
        centroids_t=bundle["head_t"].centroids_,
        thresholds=bundle["thresholds"],
        scaler_mean=bundle["pre"].scaler_.mean_,
        scaler_scale=bundle["pre"].scaler_.scale_,
        median=bundle["pre"].median_.to_dict(),
        config=CFG,
    ), path)
    print(f"[{bundle['tag']}] weights + preprocessing state saved -> {path}")
    return path


# %% =====================  CELL 15 — RUN EVERYTHING  ========================
banner("CELL 15 — FULL PIPELINE EXECUTION")

CV, BASE, FINAL, HOLD, IMP = {}, {}, {}, {}, {}

for key, d in COHORTS.items():
    label = d["label"]
    banner(f">>> {label}", "-")
    sp = SPLITS[key]

    CV[key] = run_cv(sp["X_dev"], sp["y_dev"], key)
    for m in CV[key]["models"]:
        print(f"\n[{key}] {m} — 5-fold CV (mean ± std)")
        print(CV[key]["summary"][m].loc[METRIC_ORDER].round(4).to_string())
    print(f"\n[{key}] kNN-in-embedding-space sanity check (accuracy): " +
          ", ".join(f"{m}={np.mean(v):.3f}" for m, v in CV[key]["knn_acc"].items()))
    print(f"[{key}] mean validation-fold embedding silhouette: " +
          ", ".join(f"{m}={np.nanmean(v):.3f}"
                    for m, v in CV[key]["silhouette"].items()))

    visualise_cv(CV[key], label)
    BASE[key] = run_baselines(sp["X_dev"], sp["y_dev"], key)

    FINAL[key] = fit_final(sp["X_dev"], sp["y_dev"], key, verbose=False)
    HOLD[key] = verify_holdout(FINAL[key], sp["X_hold"], sp["y_hold"], label)
    IMP[key] = permutation_importance_bundle(FINAL[key], sp["X_hold"],
                                             sp["y_hold"], label)
    save_bundle(FINAL[key])


# %% ===========  CELL 16 — CROSS-MODEL COMPARISON + FINAL REPORT  ===========
banner("CELL 16 — FINAL COMPARISON REPORT")

for key in COHORTS:
    label = COHORTS[key]["label"]
    deep = {m: CV[key]["summary"][m]["mean"] for m in CV[key]["models"]}
    base = {m: BASE[key]["summary"][m]["mean"] for m in BASE[key]["summary"]}
    comp = pd.DataFrame({**deep, **base}).loc[METRIC_ORDER]

    print(f"\n### {label} — 5-fold CV, deep models vs classical baselines")
    print(comp.round(3).to_string())

    plt.figure(figsize=(11, 4.4))
    sub = comp.loc[["Balanced_Accuracy", "F1", "MCC", "ROC_AUC", "PR_AUC"]]
    sub.plot(kind="bar", ax=plt.gca(), width=.8, colormap="tab10")
    plt.ylim(0, 1.05); plt.ylabel("score"); plt.xticks(rotation=0)
    plt.title(f"{label}\n5-fold CV: Siamese / Triplet / CapsNet vs baselines")
    plt.legend(fontsize=7, ncol=2)
    savefig(f"{key}_model_comparison")

    hold = HOLD[key]["table"].loc[METRIC_ORDER]
    plt.figure(figsize=(7.2, 7.4))
    sns.heatmap(hold.astype(float), annot=True, fmt=".3f", cmap="Greens",
                cbar_kws=dict(shrink=.6), annot_kws=dict(fontsize=8))
    plt.title(f"{label}\nUNSEEN hold-out metrics")
    savefig(f"{key}_holdout_metric_heatmap")

    cv_vs_hold = pd.DataFrame({
        "CV_mean": comp.loc[["Balanced_Accuracy", "F1", "MCC", "ROC_AUC"],
                            ["Siamese", "Triplet", "CapsNet"]].stack(),
        "Holdout": hold.loc[["Balanced_Accuracy", "F1", "MCC", "ROC_AUC"],
                            ["Siamese", "Triplet", "CapsNet"]].astype(float).stack(),
    })
    cv_vs_hold["gap"] = cv_vs_hold["CV_mean"] - cv_vs_hold["Holdout"]
    print(f"\n{label} — generalisation check (CV mean vs unseen hold-out). "
          f"A large positive gap = the model memorised the development set:")
    print(cv_vs_hold.round(3).to_string())

    comp.round(4).to_csv(os.path.join(OUTDIR, f"{key}_cv_metrics.csv"))
    hold.round(4).to_csv(os.path.join(OUTDIR, f"{key}_holdout_metrics.csv"))
    IMP[key].round(5).to_csv(os.path.join(OUTDIR, f"{key}_perm_importance.csv"))

    best = comp.loc["Balanced_Accuracy", ["Siamese", "Triplet", "CapsNet"]].idxmax()
    print(f"\n  best deep model on {key} by CV balanced accuracy: {best}")
    lr = comp.loc["Balanced_Accuracy", "LogisticRegression"]
    bd = comp.loc["Balanced_Accuracy", best]
    verdict = ("the deep metric/capsule models beat it"
               if bd > lr else "logistic regression matches or beats them — "
               "on this sample size the extra complexity is not paying for itself")
    print(f"  logistic-regression balanced accuracy = {lr:.3f}, "
          f"{best} = {bd:.3f} -> {verdict}")

# ---------------------- brand-new-patient demonstration ---------------------
banner("NEW-PATIENT PREDICTION DEMO (raw values in, probability out)")

demo_A = [
    # a profile matching the classic PCOS presentation
    dict(**{" Age (yrs)": 26, "Weight (Kg)": 82, "Height(Cm)": 158, "BMI": 32.8,
            "Cycle(R/I)": 4, "Cycle length(days)": 2, "AMH(ng/mL)": 9.5,
            "Follicle No. (L)": 15, "Follicle No. (R)": 16, "Follicle_total": 31,
            "Weight gain(Y/N)": 1, "hair growth(Y/N)": 1,
            "Skin darkening (Y/N)": 1, "Pimples(Y/N)": 1,
            "Fast food (Y/N)": 1, "Reg.Exercise(Y/N)": 0}),
    # a profile matching a healthy control
    dict(**{" Age (yrs)": 31, "Weight (Kg)": 54, "Height(Cm)": 163, "BMI": 20.3,
            "Cycle(R/I)": 2, "Cycle length(days)": 5, "AMH(ng/mL)": 2.1,
            "Follicle No. (L)": 4, "Follicle No. (R)": 5, "Follicle_total": 9,
            "Weight gain(Y/N)": 0, "hair growth(Y/N)": 0,
            "Skin darkening (Y/N)": 0, "Pimples(Y/N)": 0,
            "Fast food (Y/N)": 0, "Reg.Exercise(Y/N)": 1}),
]
demo_A = [{k: v for k, v in r.items() if k in X_A.columns} for r in demo_A]
pred_demo_A = predict_new_patients(FINAL["A_clinical"], demo_A, X_A)

demo_B = [
    dict(Age=2, Weight_kg=78.0, Height_cm=157.0, BMI_derived=31.6,
         Menstrual_Irregularity=1.0, Hormonal_Imbalance=1.0, Hirsutism=1.0,
         Hyperandrogenism=1.0, Insulin_Resistance=1.0, Family_History_PCOS=1.0,
         Exercise_Frequency=0, Sleep_Hours=0, Stress_Level=1,
         Diet_Fried_Food=6, Diet_Sweets=6, Diet_Vegetables=1),
    dict(Age=1, Weight_kg=52.0, Height_cm=165.0, BMI_derived=19.1,
         Menstrual_Irregularity=0.0, Hormonal_Imbalance=0.0, Hirsutism=0.0,
         Hyperandrogenism=0.0, Insulin_Resistance=0.0, Family_History_PCOS=0.0,
         Exercise_Frequency=4, Sleep_Hours=1, Stress_Level=0,
         Diet_Fried_Food=1, Diet_Sweets=1, Diet_Vegetables=6),
]
demo_B = [{k: v for k, v in r.items() if k in X_B.columns} for r in demo_B]
pred_demo_B = predict_new_patients(FINAL["B_lifestyle"], demo_B, X_B)

banner("DONE")
print(f"figures : {_FIG_N[0]} PNGs in ./{FIGDIR}/")
print(f"outputs : metric CSVs + model weights in ./{OUTDIR}/")
print("""
HOW TO READ THIS RUN
--------------------
1. Cohort A (clinical hormone/ultrasound panel) is the dataset where these
   models have real signal. Cohort B (lifestyle survey, n=172, 38 positives)
   is small and noisy — treat its numbers as indicative, not conclusive, and
   note the wide per-fold std.
2. Compare the CV column against the HOLD-OUT column for the same model. Close
   values = the model generalises. A big drop = it memorised the folds.
3. Always read Balanced_Accuracy / MCC / PR-AUC before Accuracy on cohort B:
   a model that predicts "No PCOS" for everyone already scores ~0.78 accuracy.
4. The baselines in CELL 12 are the sanity check. If logistic regression is
   level with the Siamese/Triplet/CapsNet numbers, that is the honest finding
   for a dataset of this size, not a bug to tune away.
""")
```
